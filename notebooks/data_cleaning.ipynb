```python
import pandas as pd
```

```python
df = pd.read_csv('/content/sample_data/credit_card_transactions.csv')
```

```python
initial_rows = len(df)
df = df.drop_duplicates(subset=['trans_num'], keep='first')
print(f"Removed {initial_rows - len(df)} duplicates")

df = df.dropna(subset=['trans_num', 'amt', 'trans_date_trans_time'])

if 'merchant' in df.columns:
    df['merchant'] = df['merchant'].fillna('Unknown')

for col in ['category']:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].mode()[0] if not df[col].mode().empty else 'Missing')

df['trans_date_trans_time'] = pd.to_datetime(df['trans_date_trans_time'], errors='coerce')

df = df.dropna(subset=['trans_date_trans_time'])

df['amt'] = df['amt'].astype(str).str.replace('[\$,]', '', regex=True)
df['amt'] = pd.to_numeric(df['amt'], errors='coerce')
df = df.dropna(subset=['amt'])

df = df[df['amt'] > 0]

amt_mean = df['amt'].mean()
amt_std = df['amt'].std()
df['amount_outlier'] = (df['amt'] > amt_mean + 3*amt_std) | (df['amt'] < amt_mean - 3*amt_std)

df['year'] = df['trans_date_trans_time'].dt.year
df['month'] = df['trans_date_trans_time'].dt.month
df['day_of_week'] = df['trans_date_trans_time'].dt.day_name()
df['hour'] = df['trans_date_trans_time'].dt.hour

text_cols = ['category', 'gender']
for col in text_cols:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip().str.upper()

print("\n--- After Cleaning ---")
print(f"Final shape: {df.shape}")
print(f"Missing amounts: {df['amt'].isnull().sum()}")
print(f"Outlier transactions flagged: {df['amount_outlier'].sum()}")
print("\nData types:\n", df.dtypes)
```

```python
df.to_csv('credit_card_cleaned.csv', index=False)
print("Cleaned data saved to 'credit_card_cleaned.csv'")
```
